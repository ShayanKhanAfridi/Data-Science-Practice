# **Web Scraping**

**Web Scraping** is the process of automatically extracting data from websites using programs or scripts.

It allows us to collect information such as text, images, prices, or tables from web pages and store it in a structured format like CSV or a database.

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
webpage = requests.get("https://oladoc.com/pakistan/karachi/hospital").text  # with this we get the complete html souce code of the webpage
```

```python
soup = BeautifulSoup(webpage, 'lxml')
```

```python
# print(soup.prettify())
```

```python
hospitals = soup.find_all('div', class_='card-body p-2 py-lg-3')    # har hospital ke div me ja rahe hain, phir loop laga kar individually sabka data nikalenge
len(hospitals)
```

```python
name = []
address = []
contact = []
panel_drs = []
area = []
timings = []

for i in hospitals:

    # Hospital Name
    h2_tag = i.find('h2')
    name.append(h2_tag.text.strip() if h2_tag else None)

    # Address
    address_tag = i.find('small', class_='text-muted text-sm')
    address.append(address_tag.text.strip() if address_tag else None)

    # Contact
    phone_tag = i.find('a', id='call-tooltip-opner')
    if phone_tag and phone_tag.get('href'):
        contact.append(phone_tag['href'].replace('tel:', '').strip())
    else:
        contact.append(None)

    # Panel Doctors
    panel_tag = i.find('small', class_='col pl-0')
    panel_drs.append(panel_tag.text.strip() if panel_tag else None)

    # Area & Timings (same class — so use find_all)
    info_tags = i.find_all('small', class_='col pl-lg-0')

    if len(info_tags) >= 2:
      timings.append(info_tags[0].text.strip())
      area.append(info_tags[1].text.strip())
    else:
      area.append(None)
      timings.append(None)


# Create DataFrame
d = {
    "Hospital Name": name,
    "Address": address,
    "Contact": contact,
    "Panel Doctors": panel_drs,
    "Area": area,
    "Timings": timings
}

df = pd.DataFrame(d)
```

## **get all pages data**

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup

headers = {
    "User-Agent": "Mozilla/5.0"
}

final = pd.DataFrame()

for page in range(0, 1470, 10):
    url = f"https://oladoc.com/pakistan/karachi/hospital/{page}"

    webpage = requests.get(url, headers=headers).text
    soup = BeautifulSoup(webpage, 'lxml')
    hospitals = soup.find_all('div', class_='card-body p-2 py-lg-3')

    name = []
    address = []
    contact = []
    panel_drs = []
    area = []
    timings = []

    for i in hospitals:

        h2_tag = i.find('h2')
        name.append(h2_tag.text.strip() if h2_tag else None)

        address_tag = i.find('small', class_='text-muted text-sm')
        address.append(address_tag.text.strip() if address_tag else None)

        phone_tag = i.find('a', id='call-tooltip-opner')
        if phone_tag and phone_tag.get('href'):
            contact.append(phone_tag['href'].replace('tel:', '').strip())
        else:
            contact.append(None)

        panel_tag = i.find('small', class_='col pl-0')
        panel_drs.append(panel_tag.text.strip() if panel_tag else None)

        info_tags = i.find_all('small', class_='col pl-lg-0')

        if len(info_tags) >= 2:
            area.append(info_tags[0].text.strip())
            timings.append(info_tags[1].text.strip())
        else:
            area.append(None)
            timings.append(None)

    df = pd.DataFrame({
        "Hospital Name": name,
        "Address": address,
        "Contact": contact,
        "Panel Doctors": panel_drs,
        "Area": area,
        "Timings": timings
    })

    final = pd.concat([final, df], ignore_index=True)

print("Total records:", len(final))
```
